### Objective
To analyze the distribution of Movies and TV Shows available in the Netflix dataset.

```python
import pandas as pd
```

```python
cleaned_df = pd.read_csv("Netflix_Cleaned_Dataset.csv")

cleaned_df.head()
```

```python
content_count = cleaned_df['type'].value_counts()

print("Movies and TV Shows Count:")
print(content_count)
```

```python
import matplotlib.pyplot as plt

content_count.plot(kind='bar')

plt.title('Netflix Content Distribution')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(6, 6))

plt.pie(
    content_count,
    labels=content_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage of Movies and TV Shows on Netflix')
plt.show()
```

## Key Findings

- The dataset contains a total of 8,790 Netflix titles.
- Movies account for 6,126 titles, representing approximately 69.7% of the total content.
- TV Shows account for 2,664 titles, representing approximately 30.3% of the total content.
- Movies are the dominant content type in the dataset.
- The number of Movies is more than twice the number of TV Shows.
